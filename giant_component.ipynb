$$S = 1 - \frac{p^2}{3(1-p)} - \frac{p^3}{27(1-p)^2}, \quad p<\tfrac34,\qquad S=0 \text{ otherwise.}$$

```python
import numpy as np
import matplotlib.pyplot as plt

def G(x, p): return p*x + (1-p)*x**3
def H(x, p): return (p + 3*(1-p)*x**2) / (3 - 2*p)

def u_of_p(p):
    p = np.asarray(p, dtype=float)
    return np.where(p < 0.75, p/(3*(1-p)), 1.0)

def S_of_p(p):
    return 1 - G(u_of_p(p), p)

p = np.linspace(0, 0.999, 1000)
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(p, S_of_p(p), lw=2, label=r"$S(p_1)$")
ax.set_xlabel(r"$p_1$"); ax.set_ylabel("Giant component $S$")
ax.set_title("Giant component size vs $p_1$ ")
ax.set_xlim(0, 1); ax.set_ylim(0, 1.02)
ax.grid(alpha=.3); ax.legend()
plt.show()
```
